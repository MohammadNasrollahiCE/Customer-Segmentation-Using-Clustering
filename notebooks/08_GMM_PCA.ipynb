#### GMM with PCA

**Gaussian Mixture Model** was evaluated on **different numbers of principal components**. The best configuration was obtained using the **first principal component (PC0)**, producing two clusters with a Silhouette Score of approximately **0.60**.

The two clusters were clearly separated along PC0, with mean PC0 values of approximately 1.65 and -1.13. The resulting cluster profiles were also **highly similar to those obtained using K-Means**, indicating that the identified `two-cluster structure` is not `strongly dependent on the choice of clustering algorithm.`

The first principal component was `mainly` associated with **Grocery, Milk, and Detergents_Paper**, with `opposite contributions` from **Fresh and Frozen**. Therefore, the two clusters can be interpreted primarily as a contrast between `customers with relatively higher Grocery/Milk/Detergents_Paper spending` and `customers with relatively higher Fresh/Frozen spending`.

#### Cluster profile with Z-Score, it shows the different between cluster mean and overal mean for any features

```python
df_Z = pd.DataFrame(index=sorted(df_exp['labels'].unique()))

for feature in features:
    cluster_mean = df_exp.groupby('labels')[feature].mean()
    overall_mean = df_exp[feature].mean()
    overall_std = df_exp[feature].std()
    
    df_Z[feature] = (cluster_mean - overall_mean) / overall_std

df_Z
```

#### Shwoing the pattern using heatmap

```python
plt.figure(figsize= (6 , 4))

sns.heatmap(data= df_Z, annot= True, fmt= '.2f')

plt.show()
```

## Conclusion

After evaluating the **Gaussian Mixture Model (GMM)** with different numbers of **Principal Components**, a clear two-cluster structure was identified when using the **first principal component (PC0)** alone. The best configuration was obtained with **K = 2**, achieving a Silhouette Score of approximately **0.60**.

This result is substantially higher than the best Silhouette Score obtained on the original standardized six-dimensional feature space, and is very close to the result obtained with **K-Means using PC0**. The fact that GMM and K-Means both identify a strong two-cluster structure in the same reduced space provides additional evidence that the observed segmentation is not strongly dependent on a particular clustering algorithm.

The cluster assignments were also relatively stable in terms of their overall structure. GMM assigned **161 customers to Cluster 0** and **234 customers to Cluster 1**, resulting in a reasonably balanced segmentation rather than a solution dominated by a very small cluster.

The first principal component (PC0) is primarily associated with **Grocery, Milk, and Detergents_Paper**, while showing opposite contributions from **Fresh and Frozen**. Therefore, PC0 can be interpreted as a dominant **purchasing-pattern axis**, separating customers according to their relative purchasing behavior across these product categories.

The resulting cluster distributions along PC0 support this interpretation. Cluster 0 has a positive mean PC0 value of approximately **1.65**, while Cluster 1 has a negative mean value of approximately **-1.13**. This indicates that the two groups occupy different regions of the dominant purchasing-pattern axis, with Cluster 0 generally associated with higher values of the variables contributing positively to PC0.

The cluster profiles in the original standardized feature space further reveal the characteristics of the two segments:

| Cluster |  Fresh |   Milk | Grocery | Frozen | Detergents_Paper | Delicassen |
| ------- | -----: | -----: | ------: | -----: | ---------------: | ---------: |
| 0       | -0.187 |  0.866 |   0.938 | -0.230 |            0.925 |      0.376 |
| 1       |  0.129 | -0.596 |  -0.646 |  0.159 |           -0.636 |     -0.259 |

Cluster 0 is characterized by substantially higher standardized purchasing levels for **Milk, Grocery, and Detergents_Paper**, together with a moderately higher value for **Delicassen**. In contrast, Cluster 1 shows relatively higher values for **Fresh and Frozen**, while having substantially lower values for **Milk, Grocery, and Detergents_Paper**.

The largest differences between the two clusters occur in **Grocery, Milk, and Detergents_Paper**, which is consistent with the loadings of PC0 and confirms that these variables play a major role in defining the segmentation. The differences in **Fresh and Frozen** are smaller in magnitude but occur in the opposite direction, further supporting the interpretation of PC0 as a contrast between these purchasing patterns.

The distribution of customers across **Region** is relatively similar between the two clusters:

| Cluster | Region 1 | Region 2 | Region 3 |
| ------- | -------: | -------: | -------: |
| 0       |   16.77% |    9.94% |   73.29% |
| 1       |   19.66% |   11.54% |   68.80% |

Both clusters are dominated by **Region 3**, and the regional distributions are broadly similar. Therefore, Region does not appear to be the primary factor distinguishing the two clusters.

A much stronger difference is observed in the **Channel** distribution:

| Cluster | Channel 1 | Channel 2 |
| ------- | --------: | --------: |
| 0       |    29.81% |    70.19% |
| 1       |    96.15% |     3.85% |

Cluster 0 contains customers predominantly associated with **Channel 2**, whereas Cluster 1 is overwhelmingly associated with **Channel 1**. This substantial difference suggests that the identified purchasing-pattern segments are also strongly related to the customers' channel characteristics. However, because Channel was not used as an input feature for the clustering, this difference should be interpreted as an **external characteristic of the resulting clusters**, rather than as a variable directly responsible for their formation.

Overall, the GMM analysis indicates that the Wholesale Customers dataset contains a **clear and reasonably stable two-cluster structure**, with the strongest separation captured by **PC0**. The similarity between the GMM and K-Means solutions, both in terms of Silhouette Score and cluster profiles, provides additional support for the presence of this underlying segmentation.

The resulting segments can primarily be interpreted as a contrast between **customers with relatively higher Grocery, Milk, and Detergents_Paper purchasing** and **customers with relatively higher Fresh and Frozen purchasing**. The strong difference in Channel distributions provides an additional business-level characteristic of these segments, while the relatively similar Region distributions suggest that geographic region plays a smaller role in distinguishing them.

Therefore, **GMM with two clusters using PC0** provides a meaningful alternative clustering solution and independently reproduces the dominant two-segment structure identified by K-Means. Further analysis can focus on comparing the GMM and K-Means assignments directly, examining GMM posterior probabilities and uncertainty, and developing a more detailed business interpretation of the resulting customer segments.
